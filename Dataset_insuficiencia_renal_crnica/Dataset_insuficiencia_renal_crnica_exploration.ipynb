```python
df = df.rename(columns={
    "Identificación del paciente": "Patient_ID",
    "Edad": "Age",
    "Presión arterial": "Blood_Pressure",
    "Gravedad específica de la orina": "Specific_Gravity",
    "Albúmina en orina": "Albumin",
    "Azúcar en orina": "Sugar",
    "Glóbulos rojos": "Red_Blood_Cells",
    "Células de pus": "Pus_Cells",
    "Grupos de células de pus": "Pus_Cell_Clumps",
    "Bacterias": "Bacteria",
    "Glucosa en sangre aleatoria": "Blood_Glucose_Random",
    "Urea en sangre": "Blood_Urea",
    "Creatinina sérica": "Serum_Creatinine",
    "Sodio en sangre": "Sodium",
    "Potasio en sangre": "Potassium",
    "Hemoglobina": "Hemoglobin",
    "Volumen corpuscular empaquetado": "Packed_Cell_Volume",
    "Recuento de glóbulos blancos": "White_Blood_Cell_Count",
    "Recuento de glóbulos rojos": "Red_Blood_Cell_Count",
    "Apetito": "Appetite",
    "Insuficiencia renal crónica": "Chronic_Kidney_Disease"
})
```

```python
identifier_columns = ["Patient_ID","Age"]

numerical_columns = ["Age","Blood_Pressure", "Specific_Gravity", "Albumin", "Sugar", "Blood_Glucose_Random",
       "Blood_Urea", "Serum_Creatinine", "Sodium", "Potassium", "Hemoglobin", "Packed_Cell_Volume",
       "White_Blood_Cell_Count", "Red_Blood_Cell_Count"] 

category_columns = ["Red_Blood_Cells", "Pus_Cells", "Pus_Cell_Clumps", "Bacteria", "Appetite"]

target_column = "Chronic_Kidney_Disease"

```

```python

print("Values of categorical columns")
for column in category_columns:
    print(df[column].unique())
    

print("Count of unique values in categorical columns")
for column in category_columns:
    print(df[column].value_counts(dropna=False))

```

```python
older_than_40 = df["Age"] > 40

display(df.loc[older_than_40, ["Age", "Blood_Pressure"]].head())
```

```python
mask = (df["Age"] >= 30) & (df["Blood_Pressure"] > 80)

display(df.loc[mask, ["Age", "Blood_Pressure"]].head())
```

```python
mask = ~(df["Age"] > 40)

display(df.loc[mask, ["Age"]].head())
```
