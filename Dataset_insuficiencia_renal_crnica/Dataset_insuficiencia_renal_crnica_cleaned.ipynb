```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
df = pd.read_csv("data/Dataset_insuficiencia_renal_crnica.csv")
clean_df = df.copy()
```

```python
clean_df = clean_df.rename(columns={
    "Identificación del paciente": "Patient_ID",
    "Edad": "Age",
    "Presión arterial": "Blood_Pressure",
    "Gravedad específica de la orina": "Specific_Gravity",
    "Albúmina en orina": "Albumin",
    "Azúcar en orina": "Sugar",
    "Glóbulos rojos": "Red_Blood_Cells",
    "Células de pus": "Pus_Cells",
    "Grupos de células de pus": "Pus_Cell_Clumps",
    "Bacterias": "Bacteria",
    "Glucosa en sangre aleatoria": "Blood_Glucose_Random",
    "Urea en sangre": "Blood_Urea",
    "Creatinina sérica": "Serum_Creatinine",
    "Sodio en sangre": "Sodium",
    "Potasio en sangre": "Potassium",
    "Hemoglobina": "Hemoglobin",
    "Volumen corpuscular empaquetado": "Packed_Cell_Volume",
    "Recuento de glóbulos blancos": "White_Blood_Cell_Count",
    "Recuento de glóbulos rojos": "Red_Blood_Cell_Count",
    "Apetito": "Appetite",
    "Insuficiencia renal crónica": "Chronic_Kidney_Disease"
})

print(clean_df.columns.tolist())
```

```python
numeric_df = clean_df.select_dtypes(include="number")
outlier_results = []

for col in numeric_df.columns:
    Q1 = numeric_df[col].quantile(0.25)
    Q3 = numeric_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers_count = (
        (numeric_df[col] < lower) | (numeric_df[col] > upper)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Outliers Count": int(outliers_count)
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

```python
checks = {
    "Age": (2, 90),
    "Blood_Pressure": (50, 180),
    "Specific_Gravity": (1.005, 1.025),
    "Albumin": (0, 5),
    "Sugar": (0, 5),
    "Blood_Glucose_Random": (22, 490),
    "Blood_Urea": (1.5, 391),
    "Serum_Creatinine": (0.4, 76),
    "Hemoglobin": (3.1, 17.8),
    "Packed_Cell_Volume": (9, 54),
    "White_Blood_Cell_Count": (2200, 26400),
    "Red_Blood_Cell_Count": (2.1, 8)
}

for col, (low, high) in checks.items():
    if col in clean_df.columns:
        values = pd.to_numeric(clean_df[col], errors="coerce")
        mask = values.notna() & ~values.between(low, high)

        print(f"\n{col}: {mask.sum()} values outside reference dataset range")
        display(clean_df.loc[mask, [col]].head(10))
```

```python
clean_df = clean_df.dropna(subset=["Chronic_Kidney_Disease"]).copy()

print("shape after removing missing target values:", clean_df.shape)
print("missing target values:", clean_df["Chronic_Kidney_Disease"].isna().sum())
```

feature engineering

```python
clean_df["Age_Group"] = pd.cut(
    clean_df["Age"],
    bins=[-np.inf, 18, 40, 60, np.inf],
    labels=["child", "young_adult", "middle_aged", "senior"],
    right=False
)

clean_df["Blood_Pressure_Category"] = pd.cut(
    clean_df["Blood_Pressure"],
    bins=[-np.inf, 90, 120, 140, np.inf],
    labels=["low", "normal", "elevated", "high"],
    right=False
)

clean_df["Blood_Glucose_Category"] = pd.cut(
    clean_df["Blood_Glucose_Random"],
    bins=[-np.inf, 70, 140, 200, np.inf],
    labels=["low", "normal", "elevated", "high"],
    right=False
)

print(clean_df[[
    "Age",
    "Age_Group",
    "Blood_Pressure",
    "Blood_Pressure_Category",
    "Blood_Glucose_Random",
    "Blood_Glucose_Category"
]].head())

print("shape after feature engineering:", clean_df.shape)

print(clean_df[[
    "Age_Group",
    "Blood_Pressure_Category",
    "Blood_Glucose_Category"
]].isna().sum())
```
