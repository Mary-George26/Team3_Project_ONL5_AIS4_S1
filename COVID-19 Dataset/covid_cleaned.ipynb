```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
import csv
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
import sys
np.set_printoptions(threshold=sys.maxsize)
from IPython.display import display
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")

df = pd.read_csv(
    "Covid Data.csv",
    sep=",",
    engine="python",
    quoting=csv.QUOTE_MINIMAL
)

df_clean = df.copy(deep=True)
```

```python
df_clean["DATE_DIED_original"] = df_clean["DATE_DIED"].copy()

df_clean["NO_DEATH_DATE"] = df_clean["DATE_DIED"].eq("9999-99-99")

df_clean["DATE_DIED"] = pd.to_datetime(
    df_clean["DATE_DIED"].replace("9999-99-99", pd.NA),
    format="%d/%m/%Y",
    errors="coerce"
)

print("New DATE_DIED dtype:", df_clean["DATE_DIED"].dtype)
print("Missing dates:", df_clean["DATE_DIED"].isna().sum())
print("No-death-date marker:", df_clean["NO_DEATH_DATE"].sum())
print("Unparsed dates:",
      (df_clean["DATE_DIED"].isna() & ~df_clean["NO_DEATH_DATE"]).sum())
```

```python
special_codes = {
    "INTUBED": [97, 99],
    "ICU": [97, 99],
    "PREGNANT": [97, 98],
    "PNEUMONIA": [99],
    "DIABETES": [98],
    "COPD": [98],
    "ASTHMA": [98],
    "INMSUPR": [98],
    "HIPERTENSION": [98],
    "OTHER_DISEASE": [98],
    "CARDIOVASCULAR": [98],
    "OBESITY": [98],
    "RENAL_CHRONIC": [98],
    "TOBACCO": [98]
}


for col, codes in special_codes.items():
    df_clean[f"{col}_NOT_APPLICABLE"] = df_clean[col].eq(97)
    df_clean[f"{col}_UNKNOWN"] = df_clean[col].isin([98, 99])


for col, codes in special_codes.items():
    df_clean[col] = df_clean[col].mask(
        df_clean[col].isin(codes), pd.NA
    ).astype("Int64")


print("Dataset shape:", df_clean.shape)

for col in special_codes:
    print(
        col,
        "| Missing:", df_clean[col].isna().sum(),
        "| Remaining values:",
        sorted(df_clean[col].dropna().unique().tolist())
    )
```

```python
numeric_cols = ["AGE"]
date_cols = ["DATE_DIED"]


print("=== DESCRIPTIVE STATISTICS ===")
display(df_clean[numeric_cols].describe().T)

outlier_summary = []

for col in numeric_cols:
    s = pd.to_numeric(df_clean[col], errors="coerce").dropna()

    Q1 = s.quantile(0.25)
    Q2 = s.quantile(0.50)
    Q3 = s.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Median (Q2)": Q2,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Below Lower Bound": (s < lower_bound).sum(),
        "Above Upper Bound": (s > upper_bound).sum(),
        "Total Outliers": (
            (s < lower_bound) | (s > upper_bound)
        ).sum()
    })

outlier_summary = pd.DataFrame(outlier_summary)

print("\n=== QUARTILES, IQR & OUTLIERS ===")
display(outlier_summary)
```

```python
age_outliers = df_clean.loc[df_clean["AGE"] > 87.5, "AGE"]

print("Number of age outliers:", len(age_outliers))

print("\nAge frequencies:")
display(age_outliers.value_counts().sort_index().to_frame("Count"))

print("\nSelected percentiles:")
display(
    df_clean["AGE"].quantile(
        [0.90, 0.95, 0.99, 0.995, 0.999]
    ).to_frame("AGE")
)

print("\nRecords with age above 100:")
display(
    df_clean.loc[df_clean["AGE"] > 100, ["AGE"]]
    .value_counts()
    .sort_index()
    .to_frame("Count")
)
```

```python
checks = {
    "AGE below 0": df_clean["AGE"] < 0,
    "AGE above 100": df_clean["AGE"] > 100,

    "SEX invalid": ~df_clean["SEX"].isin([1, 2]),
    "PATIENT_TYPE invalid": ~df_clean["PATIENT_TYPE"].isin([1, 2]),
    "USMER invalid": ~df_clean["USMER"].isin([1, 2]),
    "MEDICAL_UNIT invalid": ~df_clean["MEDICAL_UNIT"].isin(range(1, 14)),
    "CLASIFFICATION_FINAL invalid":
        ~df_clean["CLASIFFICATION_FINAL"].isin(range(1, 8)),

    "INTUBED invalid":
        df_clean["INTUBED"].notna() & ~df_clean["INTUBED"].isin([1, 2]),
    "ICU invalid":
        df_clean["ICU"].notna() & ~df_clean["ICU"].isin([1, 2]),
    "PREGNANT invalid":
        df_clean["PREGNANT"].notna() & ~df_clean["PREGNANT"].isin([1, 2]),

    "PNEUMONIA invalid":
        df_clean["PNEUMONIA"].notna() & ~df_clean["PNEUMONIA"].isin([1, 2]),
    "DIABETES invalid":
        df_clean["DIABETES"].notna() & ~df_clean["DIABETES"].isin([1, 2]),
    "COPD invalid":
        df_clean["COPD"].notna() & ~df_clean["COPD"].isin([1, 2]),
    "ASTHMA invalid":
        df_clean["ASTHMA"].notna() & ~df_clean["ASTHMA"].isin([1, 2]),
    "INMSUPR invalid":
        df_clean["INMSUPR"].notna() & ~df_clean["INMSUPR"].isin([1, 2]),
    "HIPERTENSION invalid":
        df_clean["HIPERTENSION"].notna() & ~df_clean["HIPERTENSION"].isin([1, 2]),
    "OTHER_DISEASE invalid":
        df_clean["OTHER_DISEASE"].notna() & ~df_clean["OTHER_DISEASE"].isin([1, 2]),
    "CARDIOVASCULAR invalid":
        df_clean["CARDIOVASCULAR"].notna() & ~df_clean["CARDIOVASCULAR"].isin([1, 2]),
    "OBESITY invalid":
        df_clean["OBESITY"].notna() & ~df_clean["OBESITY"].isin([1, 2]),
    "RENAL_CHRONIC invalid":
        df_clean["RENAL_CHRONIC"].notna() & ~df_clean["RENAL_CHRONIC"].isin([1, 2]),
    "TOBACCO invalid":
        df_clean["TOBACCO"].notna() & ~df_clean["TOBACCO"].isin([1, 2]),
}

logical_report = pd.DataFrame({
    "Check": checks.keys(),
    "Problematic records": [int(mask.sum()) for mask in checks.values()]
})

display(logical_report)
```
