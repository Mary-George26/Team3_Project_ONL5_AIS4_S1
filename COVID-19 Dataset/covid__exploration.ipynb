```python
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
import csv
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
import sys
np.set_printoptions(threshold=sys.maxsize)
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")

df = pd.read_csv(
    "Covid Data.csv",
    sep=",",
    engine="python",
    quoting=csv.QUOTE_MINIMAL
)

print("Dataset Shape:", df.shape)

display(df.head())

print(df.columns.tolist())

display(df.dtypes.to_frame("dtype"))

print(df.memory_usage(deep=True).sum() / 1024, "KB")

df.info()

display(df.describe().T)
```

```python
numerical_columns = ["AGE"]


binary_cols = [
    "SEX",
    "PATIENT_TYPE",
    "PNEUMONIA",
    "DIABETES",
    "COPD",
    "ASTHMA",
    "INMSUPR",
    "HIPERTENSION",
    "OTHER_DISEASE",
    "CARDIOVASCULAR",
    "OBESITY",
    "RENAL_CHRONIC",
    "TOBACCO"
]


category_cols = [
    "USMER",
    "MEDICAL_UNIT",
    "CLASIFFICATION_FINAL"
]


date_cols = ["DATE_DIED"]


special_cols = [
    "INTUBED",
    "ICU",
    "PREGNANT"
]
```

```python
display(df[numerical_columns].describe().T)


for col in binary_cols:
    display(df[col].value_counts(dropna=False).to_frame("Count"))


for col in category_cols:
    display(df[col].value_counts(dropna=False).to_frame("Count"))


for col in special_cols:
    display(df[col].value_counts(dropna=False).to_frame("Count"))


print("\nDATE_DIED sample values:")
display(df["DATE_DIED"].value_counts(dropna=False).head(10))
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": (df.isna().mean() * 100).round(2)
})

display(missing_summary)


special_codes = {
    "PNEUMONIA": [99],
    "INTUBED": [97, 99],
    "ICU": [97, 99],
    "PREGNANT": [97, 98],
    "DIABETES": [98],
    "COPD": [98],
    "ASTHMA": [98],
    "INMSUPR": [98],
    "HIPERTENSION": [98],
    "OTHER_DISEASE": [98],
    "CARDIOVASCULAR": [98],
    "OBESITY": [98],
    "RENAL_CHRONIC": [98],
    "TOBACCO": [98]
}

for col, codes in special_codes.items():
    count = df[col].isin(codes).sum()
    percentage = count / len(df) * 100

    print(
        f"{col}: {count:,} special values "
        f"({percentage:.2f}%)"
    )

print(
    "DATE_DIED = 9999-99-99:",
    (df["DATE_DIED"] == "9999-99-99").sum()
)
```

```python

duplicate_count = df.duplicated().sum()
duplicate_percentage = duplicate_count / len(df) * 100

print("Exact duplicate rows:", duplicate_count)
print("Duplicate percentage:", round(duplicate_percentage, 2), "%")

if duplicate_count > 0:
    display(df[df.duplicated(keep=False)].head(10))

unique_columns = [
    col for col in df.columns
    if df[col].is_unique
]

print("Columns unique across all rows:", unique_columns)
```

The dataset contains 812,049 exact duplicate rows (77.44% of all records). Since no unique patient identifier is available, these duplicates cannot automatically be assumed to represent repeated records of the same patients. Therefore, duplicate removal is postponed until the dataset documentation and record structure are verified. The original dataset is preserved to avoid potentially removing valid observations.

```python
print("Dataset shape:", df.shape)

print("Exact duplicate rows:", df.duplicated().sum())

print("Unique rows:", len(df.drop_duplicates()))

print("Rows duplicated beyond the first occurrence:",
      df.duplicated(keep="first").sum())
```

```python
check_cols = ["USMER", "MEDICAL_UNIT", "SEX", "PATIENT_TYPE",
              "DATE_DIED", "AGE", "CLASIFFICATION_FINAL"]

print("\nUnique combinations of selected columns:",
      len(df[check_cols].drop_duplicates()))

print("\nUnique DATE_DIED values:", df["DATE_DIED"].nunique())
print("Unique AGE values:", df["AGE"].nunique())
```

```python
print("Current dtype:", df["DATE_DIED"].dtype)

print("\nSpecial marker count:")
print((df["DATE_DIED"] == "9999-99-99").sum())

print("\nSample valid-looking date values:")
display(
    df.loc[
        df["DATE_DIED"] != "9999-99-99",
        "DATE_DIED"
    ].head(10)
)

print("\nAGE dtype:", df["AGE"].dtype)
print("Minimum age:", df["AGE"].min())
print("Maximum age:", df["AGE"].max())
```

```python
df["DATE_DIED_original"] = df["DATE_DIED"].copy()

df["NO_DEATH_DATE"] = df["DATE_DIED"].eq("9999-99-99")

df["DATE_DIED"] = pd.to_datetime(
    df["DATE_DIED"].replace("9999-99-99", pd.NA),
    format="%d/%m/%Y",
    errors="coerce"
)

print("New DATE_DIED dtype:", df["DATE_DIED"].dtype)
print("Missing dates:", df["DATE_DIED"].isna().sum())
print("No-death-date marker:", df["NO_DEATH_DATE"].sum())
print("Unparsed dates:",
      (df["DATE_DIED"].isna() & ~df["NO_DEATH_DATE"]).sum())
```
