feature engineering

```python
df_clean["bmi"] = (
    df_clean["weight"] / (df_clean["height"] / 100) ** 2
).round(2)

df_clean["bmi_category"] = pd.cut(
    df_clean["bmi"],
    bins=[0, 18.5, 25, 30, float("inf")],
    labels=["underweight", "normal", "overweight", "obese"],
    right=False
)

df_clean["pulse_pressure"] = df_clean["ap_hi"] - df_clean["ap_lo"]

df_clean["map"] = (
    (df_clean["ap_hi"] + 2 * df_clean["ap_lo"]) / 3
).round(2)

df_clean["age_group"] = pd.cut(
    df_clean["age_years"],
    bins=[0, 40, 60, float("inf")],
    labels=["young", "middle-aged", "older"],
    right=False
)

df_clean["bp_category"] = pd.cut(
    df_clean["ap_hi"],
    bins=[0, 120, 140, float("inf")],
    labels=["normal", "elevated", "high"],
    right=False
)

df_clean["cholesterol_category"] = df_clean["cholesterol"].map({
    1: "normal",
    2: "above normal",
    3: "well above normal"
})

df_clean["glucose_category"] = df_clean["gluc"].map({
    1: "normal",
    2: "above normal",
    3: "well above normal"
})

df_clean["lifestyle_risk_count"] = (
    (df_clean["smoke"] == 1).astype(int)
    + (df_clean["alco"] == 1).astype(int)
    + (df_clean["active"] == 0).astype(int)
)

df_clean["high_risk_count"] = (
    (df_clean["bmi"] >= 30).astype(int)
    + (df_clean["ap_hi"] >= 140).astype(int)
    + (df_clean["cholesterol"] >= 2).astype(int)
    + (df_clean["gluc"] >= 2).astype(int)
)

print(df_clean.columns)
print(df_clean.head())
```
