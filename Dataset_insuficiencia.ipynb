```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Dataset_insuficiencia_renal_crnica.csv')
df.head(5)
```

```python
df.shape
```

```python
df.info()
```

```python
#  stats
df.describe()
```

```python
# check nulls
df.isnull().sum()
```

```python
# check duplicates
df.duplicated().sum()
```

```python
# drop duplicates
df.drop_duplicates(inplace=True)
```

```python
# drop nulls
df.dropna(inplace=True)
```

```python
# check shape
df.shape
```

```python
# target counts
df['Insuficiencia renal crónica'].value_counts()
```

Visulization

```python
#  target
sns.countplot(data=df, x='Insuficiencia renal crónica')
plt.show()
```

```python
#  age distribution
sns.histplot(data=df, x='Edad', kde=True)
plt.show()
```

```python
#  age vs target
sns.boxplot(data=df, x='Insuficiencia renal crónica', y='Edad')
plt.show()
```

```python
# blood pressure vs target
sns.boxplot(data=df, x='Insuficiencia renal crónica', y='Presión arterial')
plt.show()
```

```python
#  creatinine vs target
sns.boxplot(data=df, x='Insuficiencia renal crónica', y='Creatinina sérica')
plt.show()
```

```python
# larger heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```
